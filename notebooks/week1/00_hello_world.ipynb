# 00 — Hello World

Smoke-test that all core dependencies are installed and the plotting backend works.

```python
import dateutil
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scipy

for name, mod in [
    ("pandas", pd),
    ("numpy", np),
    ("matplotlib", matplotlib),
    ("scipy", scipy),
    ("python-dateutil", dateutil),
]:
    version = getattr(mod, "__version__", "unknown")
    print(f"{name:>20}  {version}")
```

```python
# Trivial DataFrame
df = pd.DataFrame(
    {
        "date": pd.date_range("2024-01-01", periods=5, freq="B"),
        "close": [100.0, 101.5, 99.8, 102.3, 103.1],
    }
).set_index("date")

print(df)
print(f"\nshape: {df.shape}  |  dtype: {df.dtypes['close']}")
```

```python
# Trivial plot — confirms the backend renders without error
matplotlib.use("Agg")  # non-interactive backend; safe in any environment

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(df.index, df["close"], marker="o", linewidth=1.5)
ax.set_title("Hello World — dummy close prices")
ax.set_xlabel("Date")
ax.set_ylabel("Price")
fig.tight_layout()
plt.show()
print("Plot rendered successfully.")
```
